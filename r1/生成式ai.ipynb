# 自监督学习（Pre-training） 
生成式ai的实质是在进行一个“文字接龙”的任务。而在第一阶段的训练，就是训练师在网络上爬取海量的文本内容，然后给出前面部分，让ai预测接下来的词，再与正确答案进行对比，通过反向传播调整模型。这是训练出来的ai是一个擅长“文字接龙”的高手。其生成的内容是是基本符合语法的。但是当你向ai提问时，它往往直接接着进行续写，而非回答你的问题。这是因为此时的ai只学会了文字的模式，而没有知识。

# 监督式学习（Supervised Fine-tuning）
该阶段则是要解决ai没有知识的问题。通过人工为问题集标注答案。然后输入问题，让ai接着回答，再和正确答案比对，通过反向传播调整模型。以此使得ai学会知识。那既然如此，能否直接用这样的方式训练ai，而跳过自监督学习能？答案是否定的，因为人工标注的成本较高，数据量不足，所以需要大量无标注文本进行预训练，使得模型能基本理解人类的语言模式。在此基础上再让ai学会知识会更加高效。

# 人类反馈强化学习（RLHF）
这个阶段则是利用人类反馈，让ai学习如何通过人类反馈来进行更好的回答。这是是让人类在ai生成的多种答案中选择最好的答案，反馈给ai。其核心在于优化输出结果的整体质量。人类指出哪个答案好，模型就调整参数，提升产生好答案的概率，降低差答案的概率。使得ai的回答更加符合人类习惯。
